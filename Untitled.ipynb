```python
import pandas as pd
import matplotlib as plt
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("data/superstore.csv", encoding="latin1")
df.head()
```

```python
df["order Dtae"] =pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])
df = df.drop_duplicates()
df.to_csv("data/cleaned_superstore.csv", index=False)
```

```python
# Total revenue and profit
print("Total sales: ",df["Sales"].sum())
print("Total Profit: ",df["Profit"].sum())
```

```python
#Sales by category
category_sale = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)
category_sale.plot(kind="bar",title ="Sales by category")
plt.ylabel("Sales ($)")
plt.tight_layout()
plt.savefig("dashboard/sales_by_category.png")
plt.show()
```

```python
# Sales by region
region_profit= df.groupby("Region")["Profit"].sum().sort_values(ascending=False)
region_profit.plot(kind="bar",color="green",title="Profit by region")
plt.tight_layout()
plt.savefig("dashboard/profit_by_region.png")
plt.show()
```

```python
# Monthly sales trend
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Month"] = df["Order Date"].dt.to_period("M")
monthly_sales = df.groupby("Month")["Sales"].sum()
monthly_sales.plot(kind="line", title="Monthly Sales Trend", figsize=(10,4))
plt.tight_layout()
plt.savefig("dashboard/monthly_trend.png")
plt.show()
```

```python
# Top 10 products by sales
top_products = df.groupby("Product Name")["Sales"].sum().sort_values(ascending=False).head(10)
print(top_products)
```

```python
# Discount vs Profit relationship (classic Superstore insight — high discounts often kill profit)
sns.scatterplot(data=df, x="Discount", y="Profit")
plt.title("Discount vs Profit")
plt.savefig("dashboard/discount_vs_profit.png")
plt.show()
```
